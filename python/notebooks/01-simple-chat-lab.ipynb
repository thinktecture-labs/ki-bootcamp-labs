## Chat with system prompt

```python
# little bit more context about who is talking
from langchain_core.messages import HumanMessage, SystemMessage

# With a SystemMessage, we can establish a helpful context at the beginning of the chat. This can include background information and define the role of the AI.
# HumanMessage is the placeholder to send the user input to the AI.
# see https://reference.langchain.com/python/langchain-core/messages/human/HumanMessage for details
# Task: Use the correct combination of SystemMessage and HumanMessage to introduce yourself and then ask the AI if it recognizes your name.

result = llm_mistral.invoke([...(content="Der Name des Benutzers ist ...")] + [...(content="Kennst du meinen Namen?")])
print("--- Mistral ---")
print(result.content)
print(result.response_metadata)
print()

result = llm_qwen.invoke([...(content="Der Name des Benutzers ist ...")] + [...(content="Kennst du meinen Namen?")])
print("--- Qwen ---")
print(result.content)
print(result.response_metadata)
```
