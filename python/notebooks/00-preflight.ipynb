```python
import sys
from importlib.metadata import version

print("Python:", sys.version.split()[0])
for package in ("langchain", "langchain-openai", "langchain-qdrant", "langgraph", "langfuse"):
    print(f"{package}: {version(package)}")
print("Umgebung OK ✅")
```

```python
import os

from dotenv import load_dotenv

load_dotenv()  # lokal: .env in diesem Ordner; in Codespaces kommen die Werte aus den Secrets

required = ("OVH_API_KEY", "LANGFUSE_PUBLIC_KEY", "LANGFUSE_SECRET_KEY")
missing = [name for name in required if not os.environ.get(name)]
if missing:
    print("Fehlende Secrets ❌:", ", ".join(missing))
    print("Anlegen unter https://github.com/settings/codespaces, für das Repo freigeben,")
    print("dann den Codespace neu starten. Den OVH-Key gibt es im Workshop.")
else:
    print("Secrets OK ✅")
```

```python
from langchain_openai import OpenAIEmbeddings

if not os.environ.get("OVH_API_KEY"):
    print("OVH-Check übersprungen: OVH_API_KEY fehlt (bekommt ihr im Workshop)")
else:
    embeddings = OpenAIEmbeddings(
        model="bge-m3",
        base_url="https://oai.endpoints.kepler.ai.cloud.ovh.net/v1",
        api_key=os.environ["OVH_API_KEY"],
    )
    try:
        vector = embeddings.embed_query("Hallo Angular Days")
        print(f"OVH AI Endpoints OK ✅ ({len(vector)} Dimensionen)")
    except Exception as error:  # noqa: BLE001
        print(f"OVH AI Endpoints ❌: {error}")
```
