## Netzwerk-Check (optional)

Prüft, ob die im Workshop genutzten Dienste von hier aus erreichbar sind. Ohne API-Key ist
eine HTTP-Fehlermeldung (401/403) normal und bedeutet: erreichbar.


```python
import urllib.error
import urllib.request

checks = {
    "OVH AI Endpoints": "https://oai.endpoints.kepler.ai.cloud.ovh.net/v1/models",
    "Langfuse Cloud (EU)": "https://cloud.langfuse.com/api/public/health",
}
for name, url in checks.items():
    try:
        with urllib.request.urlopen(url, timeout=10) as response:
            print(f"{name}: erreichbar ✅ (HTTP {response.status})")
    except urllib.error.HTTPError as error:
        print(f"{name}: erreichbar ✅ (HTTP {error.code} ohne Key ist erwartet)")
    except Exception as error:  # noqa: BLE001
        print(f"{name}: NICHT erreichbar ❌ – {error}")
```
